Flujo del juego:

| Paso | Petición | Devuelve |
|------|----------|----------|
| 1 | `POST /login` | `token`, `case_id` |
| 2 | `GET /case` | datos del caso (mira `incident_time`) |
| 3 | `GET /access-log` | lista de accesos (`access_id` + hora) |
| 4 | `GET /access/{id}` | detalle del acceso sospechoso |
| 5 | `POST /case/resolve` | `{ "access_id": "..." }` → ¿correcto? |

Todos los endpoints excepto `/login` necesitan el header
`Authorization: Bearer <token>`, es decir un diccionario
`{"Authorization": f"Bearer {token}"}` pasado como `headers=...`.

## Preparación

````python
import httpx

BASE_URL = "https://stingray-monopoly-prodigal.ngrok-free.dev"
````

````python
import json


def show(respuesta: httpx.Response) -> None:
    """Imprime de forma legible el resultado de una petición."""
    print(f"{respuesta.request.method} {respuesta.url} -> {respuesta.status_code}")
    try:
        print(json.dumps(respuesta.json(), indent=2, ensure_ascii=False))
    except ValueError:
        print(respuesta.text)
````
