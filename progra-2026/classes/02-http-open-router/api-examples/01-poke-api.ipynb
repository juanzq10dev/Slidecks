# 01 — PokeAPI

API pública sobre el universo Pokémon. **No necesita API key.**

Documentación: https://pokeapi.co/docs/v2

## Preparación

Ejecuta estas dos celdas primero. `show` recibe un JSON (`respuesta.json()`) y lo imprime de forma legible.

```python
import httpx
```

```python
import json


def show(data: object) -> None:
    """Imprime un objeto JSON de forma legible."""
    print(json.dumps(data, indent=2, ensure_ascii=False))
```

## 1. Pedir un Pokémon por su nombre

`GET https://pokeapi.co/api/v2/pokemon/{nombre}`

```python
respuesta = httpx.get("https://pokeapi.co/api/v2/pokemon/charmander")
show(respuesta.json())
```

## 2. Quedarnos solo con lo que nos interesa

La respuesta es un diccionario enorme. Navegamos el JSON con claves e índices.

```python
datos = respuesta.json()

print("Nombre:", datos["name"])
print("Número:", datos["id"])
print("Altura (dm):", datos["height"])
print("Peso (hg):", datos["weight"])

tipos = [t["type"]["name"] for t in datos["types"]]
print("Tipos:", ", ".join(tipos))

habilidades = [a["ability"]["name"] for a in datos["abilities"]]
print("Habilidades:", ", ".join(habilidades))
```

```python
print("Estadísticas base:")
for stat in datos["stats"]:
    print(f"  {stat['stat']['name']:16} {stat['base_stat']}")
```

## 3. Otro endpoint: una habilidad

`GET https://pokeapi.co/api/v2/ability/{nombre}` — descripción de una habilidad.

```python
habilidad = httpx.get("https://pokeapi.co/api/v2/ability/static")
show(habilidad.json())
```
