# 02 — Rick and Morty API

Datos de personajes, ubicaciones y episodios de la serie. **No necesita API key.**

Documentación: https://rickandmortyapi.com/documentation

## Preparación

Ejecuta estas dos celdas primero. `show` imprime la respuesta de forma legible.

```python
import httpx
```

```python
import json


def show(respuesta: httpx.Response) -> None:
    """Imprime el método, la URL, el status y el JSON de la respuesta."""
    print(f"{respuesta.request.method} {respuesta.url} -> {respuesta.status_code}")
    try:
        print(json.dumps(respuesta.json(), indent=2, ensure_ascii=False))
    except ValueError:
        print(respuesta.text)
```

## 1. Un personaje por su id

`GET https://rickandmortyapi.com/api/character/{id}`

```python
respuesta = httpx.get("https://rickandmortyapi.com/api/character/148")
show(respuesta)
```

```python
respuesta_json = respuesta.json()
respuesta_json["name"]
```

## 2. Listado con paginación

`GET /character` devuelve un objeto con `info` (paginación) y `results` (lista).

```python
lista = httpx.get("https://rickandmortyapi.com/api/character")
show(lista)

```

## 3. Filtrar con query params

Pasamos un diccionario en `params=` y httpx arma la URL `?name=rick&status=alive`.

```python
respuesta = httpx.get(
    "https://rickandmortyapi.com/api/character",
    params={"name": "rick", "status": "alive"},
)
show(respuesta)
```

## 4. Seguir un enlace de la respuesta

El JSON trae URLs completas. Podemos pedir la primera ubicación de un personaje.

```python
rick_c137 = httpx.get("https://rickandmortyapi.com/api/character/1").json()
url_ubicacion = rick_c137["location"]["url"]

ubicacion = httpx.get(url_ubicacion).json()
print("Ubicación:", ubicacion["name"])
print("Tipo:", ubicacion["type"])
print("Dimensión:", ubicacion["dimension"])
```
