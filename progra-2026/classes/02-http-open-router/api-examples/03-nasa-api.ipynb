## Preparación

Ejecuta estas dos celdas primero. `show` recibe un JSON (`respuesta.json()`) y lo imprime de forma legible.

```python
import httpx
```

```python
import json


def show(data: object) -> None:
    """Imprime un objeto JSON de forma legible."""
    print(json.dumps(data, indent=2, ensure_ascii=False))
```

## 1. APOD — Foto astronómica del día

`GET https://api.nasa.gov/planetary/apod?api_key=...`

```python
respuesta = httpx.get(
    "https://api.nasa.gov/planetary/apod",
    params={"api_key": API_KEY},
)
respuesta.raise_for_status()
show(respuesta.json())
```

```python
apod = respuesta.json()
print("Título:", apod["title"])
print("Fecha:", apod["date"])
print("Imagen:", apod.get("hdurl") or apod["url"])
print()
print(apod["explanation"])
```

## 2. Pedir la foto de una fecha concreta

Agregamos el parámetro `date` (formato `AAAA-MM-DD`).

```python
respuesta = httpx.get(
    "https://api.nasa.gov/planetary/apod",
    params={"api_key": API_KEY, "date": "2020-07-04"},
)
foto = respuesta.json()
print(foto["title"], "->", foto["url"])
```

## 3. Asteroides cercanos a la Tierra (NeoWs)

`GET https://api.nasa.gov/neo/rest/v1/feed?start_date=...&end_date=...`

```python
respuesta = httpx.get(
    "https://api.nasa.gov/neo/rest/v1/feed",
    params={
        "api_key": API_KEY,
        "start_date": "2024-01-01",
        "end_date": "2024-01-02",
    },
)
datos = respuesta.json()

print("Asteroides encontrados:", datos["element_count"])

for dia, asteroides in datos["near_earth_objects"].items():
    print(f"\n{dia}:")
    for a in asteroides:
        peligroso = "⚠️ peligroso" if a["is_potentially_hazardous_asteroid"] else "ok"
        print(f"  {a['name']:25} {peligroso}")
```
