# 04 — Country API (countries.dev)

Datos de países: capital, población, monedas, idiomas, banderas...
**No necesita API key.**

Documentación: https://countries.dev

> Nota: antes usábamos `restcountries.com`, pero esa API ahora pide key.
> `countries.dev` es el reemplazo libre y sin registro.

## Preparación

Ejecuta estas dos celdas primero. `show` recibe un JSON (`respuesta.json()`) y lo imprime de forma legible.

```python
import httpx
```

```python
import json


def show(data: object) -> None:
    """Imprime un objeto JSON de forma legible."""
    print(json.dumps(data, indent=2, ensure_ascii=False))
```

## 1. Un país por su código ISO

`GET https://countries.dev/alpha/{codigo}` — código de 2 letras (`PE`, `US`, `JP`).

```python
respuesta = httpx.get("https://countries.dev/alpha/PE")
show(respuesta.json())
```

```python
pais = respuesta.json()
print("Nombre:", pais["name"])
print("Capital:", pais["capital"])
print("Región:", pais["region"], "/", pais["subregion"])
print("Población:", f'{pais["population"]:,}')
print("Bandera:", pais["flag"])

moneda = pais["currencies"][0]
print("Moneda:", moneda["name"], f'({moneda["symbol"]})')

idiomas = [i["name"] for i in pais["languages"]]
print("Idiomas:", ", ".join(idiomas))
```

## 2. Pedir solo algunos campos

Con `?fields=` reducimos el tamaño de la respuesta.

```python
respuesta = httpx.get(
    "https://countries.dev/alpha/JP",
    params={"fields": "name,capital,population,flag"},
)
show(respuesta.json())
```

## 3. Buscar por nombre

`GET https://countries.dev/name/{texto}` — devuelve una **lista** de coincidencias.

```python
resultados = httpx.get("https://countries.dev/name/republic").json()

print("Coincidencias:", len(resultados))
for pais in resultados[:10]:
    print(f"  {pais['flag']} {pais['name']:40} {pais['capital']}")
```

## 4. Comparar varios países

Hacemos una petición por cada código y juntamos los datos.

```python
codigos = ["PE", "CL", "CO", "EC", "BO"]

for codigo in codigos:
    p = httpx.get(f"https://countries.dev/alpha/{codigo}").json()
    print(f"{p['flag']} {p['name']:12} pob. {p['population']:>12,}  {p['capital']}")
```
